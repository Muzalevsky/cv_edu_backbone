## 2. Segmentation

YOLO26 _segmentation_ models use the `-seg` suffix, i.e. `yolo26n-seg.pt` and are pretrained on COCO. See [Segmentation Docs](https://docs.ultralytics.com/tasks/segment) for full details.

```
%pip install -q tensorboard

from ultralytics import YOLO, settings

# Enable TensorBoard logging
settings.update({"tensorboard": True})
```

```
# Load YOLO26n-seg, train it on COCO8-seg for 3 epochs and predict an image with it
from ultralytics import YOLO

model = YOLO('yolo26n-seg.pt')  # load a pretrained YOLO segmentation model

# train the model
model.train(data='coco8-seg.yaml', epochs=6,
    project="runs/instance",
    name="experiment_1",
    plots=True
)
```

## Visualize

```
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from PIL import Image
from matplotlib.patches import Polygon, Patch
from ultralytics.data.utils import check_det_dataset

# Get dataset paths
data = check_det_dataset("coco8-seg.yaml")

val_dir = Path(data["val"])
image_paths = sorted(
    p for p in val_dir.rglob("*")
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

class_names = model.names


def visualize_segmentation(model, image_path, conf=0.25):
    image_path = Path(image_path)

    # Load image
    image = np.array(Image.open(image_path).convert("RGB"))
    h, w = image.shape[:2]

    # Find corresponding YOLO annotation
    label_path = Path(
        str(image_path).replace("/images/", "/labels/")
    ).with_suffix(".txt")

    # Model prediction
    result = model.predict(
        str(image_path),
        conf=conf,
        verbose=False
    )[0]

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    for ax, title in zip(
        axes,
        ["Source Image", "Ground Truth", "Prediction"]
    ):
        ax.imshow(image)
        ax.set_title(title, fontsize=14)
        ax.axis("off")

    # ---------- Ground Truth ----------
    gt_legend = []
    gt_class_counts = {}

    if label_path.exists():
        with open(label_path) as f:
            for i, line in enumerate(f):
                if not line.strip():
                    continue

                values = np.array(line.split(), dtype=float)

                cls_id = int(values[0])
                polygon = values[1:].reshape(-1, 2)
                polygon = polygon * [w, h]

                class_name = class_names[cls_id]

                gt_class_counts[class_name] = (
                    gt_class_counts.get(class_name, 0) + 1
                )
                instance_id = gt_class_counts[class_name]

                label = f"{class_name} #{instance_id}"
                color = plt.cm.tab20(i % 20)

                axes[1].add_patch(
                    Polygon(
                        polygon,
                        closed=True,
                        facecolor=color,
                        edgecolor=color,
                        linewidth=2,
                        alpha=0.5
                    )
                )

                gt_legend.append(
                    Patch(
                        facecolor=color,
                        label=label
                    )
                )

    # ---------- Prediction ----------
    pred_legend = []
    pred_class_counts = {}

    if result.masks is not None:
        for i, polygon in enumerate(result.masks.xy):
            cls_id = int(result.boxes.cls[i].item())
            confidence = float(result.boxes.conf[i].item())

            class_name = class_names[cls_id]

            pred_class_counts[class_name] = (
                pred_class_counts.get(class_name, 0) + 1
            )
            instance_id = pred_class_counts[class_name]

            label = f"{class_name} #{instance_id} ({confidence:.2f})"
            color = plt.cm.tab20(i % 20)

            axes[2].add_patch(
                Polygon(
                    polygon,
                    closed=True,
                    facecolor=color,
                    edgecolor=color,
                    linewidth=2,
                    alpha=0.5
                )
            )

            pred_legend.append(
                Patch(
                    facecolor=color,
                    label=label
                )
            )

    # ---------- Legends ----------
    for ax, legend in [
        (axes[1], gt_legend),
        (axes[2], pred_legend)
    ]:
        if legend:
            ax.legend(
                handles=legend,
                loc="upper center",
                bbox_to_anchor=(0.5, -0.03),
                ncol=2,
                fontsize=9,
                frameon=False
            )

    plt.tight_layout()
    plt.show()


# Visualize validation samples
for image_path in image_paths[:4]:
    visualize_segmentation(model, image_path)
```

## 3. Semantic Segmentation

YOLO26 _semantic segmentation_ models use the `-sem` suffix, i.e. `yolo26n-sem.pt` and are pretrained on Cityscapes. See [Semantic Docs](https://docs.ultralytics.com/tasks/semantic) for full details.

```
# Load YOLO26n-sem, train it on Cityscapes8 for 3 epochs and predict an image with it
from ultralytics import YOLO

model = YOLO('yolo26n-sem.pt')  # load a pretrained YOLO semantic segmentation model
model.train(data='cityscapes8.yaml', epochs=10, project="runs/semantic",
    name="experiment_2",
    plots=True
)
```

## Visualize

```
import numpy as np
import matplotlib.pyplot as plt
import yaml

from pathlib import Path
from PIL import Image
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch

from ultralytics.data.utils import check_det_dataset
from ultralytics.utils.checks import check_yaml


# -----------------------------
# Dataset configuration
# -----------------------------

dataset_yaml = "cityscapes8.yaml"

data = check_det_dataset(dataset_yaml)
class_names = model.names

# Cityscapes masks contain original label IDs.
# Convert them to the 19 training class IDs.

raw_config = YAML.load(check_yaml(dataset_yaml))
label_mapping = {
    int(k): int(v)
    for k, v in raw_config["label_mapping"].items()
    if v != "ignore_label"
}

num_classes = len(class_names)

# One consistent color per semantic class
colors = plt.cm.tab20(np.linspace(0, 1, num_classes))
cmap = ListedColormap(colors)


def load_gt_mask(mask_path):
    """Load Cityscapes GT mask and convert label IDs to class IDs."""

    raw_mask = np.array(Image.open(mask_path))
    gt_mask = np.full(raw_mask.shape, 255, dtype=np.uint8)

    for original_id, train_id in label_mapping.items():
        gt_mask[raw_mask == original_id] = train_id

    return gt_mask


def draw_mask(ax, image, mask, alpha=0.5):
    """Draw semantic segmentation mask over the original image."""

    ax.imshow(image)

    # Ignore pixels with label 255
    masked = np.ma.masked_where(mask == 255, mask)

    ax.imshow(
        masked,
        cmap=cmap,
        vmin=0,
        vmax=num_classes - 1,
        alpha=alpha,
        interpolation="nearest"
    )

    ax.axis("off")


def visualize_semantic_segmentation(model, image_path, alpha=0.5):

    image_path = Path(image_path)

    # Load source image
    image = np.array(
        Image.open(image_path).convert("RGB")
    )

    # Corresponding GT mask
    mask_path = Path(
        str(image_path).replace("/images/", "/masks/")
    ).with_suffix(".png")

    if not mask_path.exists():
        raise FileNotFoundError(f"GT mask not found: {mask_path}")

    gt_mask = load_gt_mask(mask_path)

    # Model prediction
    result = model.predict(
        str(image_path),
        verbose=False
    )[0]

    pred_mask = result.semantic_mask.data

    if hasattr(pred_mask, "cpu"):
        pred_mask = pred_mask.cpu().numpy()

    pred_mask = np.asarray(pred_mask).squeeze()

    # Resize class IDs if output size differs from source image
    if pred_mask.shape != image.shape[:2]:
        pred_mask = np.array(
            Image.fromarray(pred_mask.astype(np.uint8)).resize(
                (image.shape[1], image.shape[0]),
                Image.Resampling.NEAREST
            )
        )

    # -----------------------------
    # Visualization
    # -----------------------------

    fig, axes = plt.subplots(1, 3, figsize=(18, 6))

    # Source
    axes[0].imshow(image)
    axes[0].set_title("Source Image", fontsize=14)
    axes[0].axis("off")

    # Ground Truth
    draw_mask(axes[1], image, gt_mask, alpha)
    axes[1].set_title("Ground Truth", fontsize=14)

    # Prediction
    draw_mask(axes[2], image, pred_mask, alpha)
    axes[2].set_title("Prediction", fontsize=14)

    # -----------------------------
    # Class legend
    # -----------------------------

    # Include classes present in GT or prediction
    present_classes = sorted(
        (set(np.unique(gt_mask)) | set(np.unique(pred_mask)))
        - {255}
    )

    legend_elements = [
        Patch(
            facecolor=colors[cls_id],
            edgecolor="black",
            label=class_names[int(cls_id)]
        )
        for cls_id in present_classes
        if 0 <= cls_id < num_classes
    ]

    fig.legend(
        handles=legend_elements,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.12),
        ncol=6,
        fontsize=10,
        frameon=False
    )

    plt.tight_layout()
    plt.show()
```

```
# Validation images
val_dir = Path(data["val"])

image_paths = sorted(
    p for p in val_dir.rglob("*")
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
)

# Display 3 samples
for image_path in image_paths[:3]:
    visualize_semantic_segmentation(
        model,
        image_path,
        alpha=0.6
    )
```
